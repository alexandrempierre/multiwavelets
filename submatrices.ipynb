```python
import numpy as np

from matplotlib import pyplot as plt

import step_4 as s4
```

```python
n, k = 8, 1
l = int(np.log2(n // k))
blocks = s4.extract_submatrices(
    np.ones((n, n)),
    k=k
)
Ts = s4.visualize_blocks(np.ones((n, n)), blocks)
```

```python
fig = plt.figure(figsize=(8, l * 8))
#
for (idx, (size, T)) in enumerate(sorted(Ts.items())):
    ax = fig.add_subplot(l, 1, idx + 1)
    ax.set_title(f'$T_{{{idx}}}$')
    ax.spy(T, marker='o', color='black')
    #
    # ax2 = fig.add_subplot(2, 1, 2)
    # ax2.set_title('$\overline{M_4}$')
    # ax2.spy(not_M4, marker=',', color='black')
    #
plt.show()
```
